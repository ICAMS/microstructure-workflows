## <font style="font-family:roboto;color:#455e6c"> Introduction </font>

A central challenge in **microstructure modeling** is that different simulation tools store and represent data in **incompatible formats**. This fragmentation makes it difficult to build **end-to-end workflows** where the output of one tool becomes the input of another. The problem becomes even more pronounced when tracking **microstructure evolution**, since most tools only record their own **internal state** and lack a **common structure** for storing **time-dependent changes** in **grains**, **phases**, and **voxel fields**.

### Integrated tools used in this notebook
- **Kanapy**: generates synthetic **3D microstructures** and performs **statistical analysis**
- **DAMASK**: performs **mechanical simulations** based on **FFT solver** with **crystal-plasticity** model
- **Pyiron_workflow**: constructs workflows as **computational graphs**

The case notebooks in this project construct and execute a complete **microstructure-to-simulation workflow**. The steps include:

- Reading and analysing an **electron backscatter diffraction (EBSD)** map
- Extracting **statistical descriptors** of the microstructure
- Generating a **3D microstructure** statistically equivalent to the **EBSD** data
- Exporting the structure to **DAMASK** as a **FAIR-DO** object
- Running a **cold-rolling simulation** in **DAMASK**
- Capturing **microstructure evolution** by updating the **FAIR-DO** object at selected **time steps**
- Re-analysing the resulted microstructure with **Kanapy** and comparing it to the **initial state**
  
### The concrete benefits of this workflow
The workflow creates a **unified**, **automated chain** from microstructure generation to mechanical simulation and post-processing. The **FAIR-DO format** ensures that results are **reusable**, **traceable**, and **preserved over time**. Automated data exchange removes **manual file handling**, and the **modular node structure** makes the workflow adaptable to a wide range of **simulation tasks**.

load_ebsd_map ─┐
               ├─> get_stats ──┐
               │               ├─> generate_rve ─> write_data ─> load_to_damask ─> run_damask ─> post_processing
  (no EBSD) ── create_stats ───┘
```

Import them with:

```python
from functions.nodes import (
    load_ebsd_map, get_stats, create_stats, generate_rve,
    write_data, load_to_damask, run_damask, post_processing,
)
```

| Node | Purpose | Key inputs | Outputs |
|---|---|---|---|
| **`load_ebsd_map`** | Reads an EBSD scan (`.ang`, `.ctf`) with Kanapy and reconstructs grains. | `file_path`, `gs_min`, `vf_min`, `max_angle`, `connectivity`, `show_plot`, `show_grains`, `felzenszwalb` | `ebsd_map` — a `knpy.EBSDmap` |
| **`get_stats`** | Extracts statistical descriptors (grain size, aspect ratio, tilt angle) **from** an EBSD map or a stats JSON. Cutoffs truncate the measured distribution to fit the target box. | `source`, `NVoxels`, `sizeRVE`, `periodic`, `deq_min/max`, `asp_min/max`, `omega_min/max` | `statisticalDescriptors` — dict, or list of dicts for multi-phase |
| **`create_stats`** | Builds the same descriptors **from scratch** (no EBSD map) for a synthetic study. | `case_type`, `phases`, `length_side`, `number_voxels`, `periodicity`, `unit` | `statisticalDescriptors` |
| **`generate_rve`** | Packs ellipsoidal grains, voxelizes them, and assigns orientations — a synthetic texture (`"goss"`, `"copper"`, `"random"`) or a real EBSD-derived ODF. | `source`, `ori`, `show_plots`, `report_grain_geometry`, `orientation_Nbase` | `RVE` — a `knpy.Microstructure` |
| **`write_data`** | Serializes the RVE into a **MiMeDO** JSON (MiMeDat schema) with user metadata, boundary conditions, phases, units and provenance; creates the `Keys/<key>/` folder tree. | `source`, `user_metadata`, `boundary_condition`, `phases`, `units`, `base_work_dir` | `MiMeDO`, `pathtoMiMeDO`, `identifier`, `key`, `input_path`, `results_path` |
| **`load_to_damask`** | Writes the three DAMASK input files from the MiMeDO object. | `source`, `grid_out`, `material_out`, `load_out`, `f_out_list`, `f_restart_list`, `outputs` | `grid`, `material`, `load` (paths) |
| **`run_damask`** | Launches the real `DAMASK_grid` spectral solver. | `grid_file`, `material_file`, `load_file`, `results_path`, `n_threads` | `results` — path to the `.hdf5` |
| **`post_processing`** | Pulls requested fields out of the DAMASK result, writes VTK, re-segments the deformed microstructure, tracks grains across increments, adds IPF colors, and writes everything back into the MiMeDO object. | `results`, `json_path`, `results_path`, `quantities`, `export_microstructure`, `export_mechanical_response`, `microstructure_stride`, `export_vtk`, `do_segmentation`, `do_grain_tracking`, `add_ipf_colors`, `do_regridding` | `UpdatedDataObject` |

> `load_to_damask`'s `mechanical_solver` and `run_damask`'s restart support (`--restart N`) are
> deliberately **not** exposed as parameters yet — see `PROJECT_STATUS.md`.

## <font style="font-family:roboto;color:#455e6c"> Identifiers and the <code>Keys/</code> folder scheme </font>

Every run gets its own folder, named by a **decomposed key** rather than one opaque hash — so
related runs are recognizable at a glance (same RVE, different loading; same loading, different texture).

```
<base_work_dir>/Keys/<key>/
├── inputs/     grid.vti, material.yaml, load.yaml
└── results/    <identifier>.json (the MiMeDO object), *.hdf5, run.log, vtk/, ...
```

```
<key> = <mimedo_identifier>_<orientation_identifier per phase>_<microstructure_state_id>
```

A real example from the Cold Rolling case:

```
Keys/fa7a100b_2e5b2_S_89e389d8/
         │        │      └── microstructure_state_id (Kanapy, "S_" prefix)
         │        └───────── orientation_identifier (per phase)
         └────────────────── mimedo_identifier
```

The scheme is modeled on [MicroMecWorkflow](https://github.com/Ronakshoghi/MicroMecWorkflow).

| Function | Where it lives | What it hashes |
|---|---|---|
| **`create_mimedo_identifier(data, hash_length=8, mandatory_fields=None)`** | `functions/mimedo_identifier.py` | The MiMeDO data dictionary, restricted to a whitelist of meaningful fields, normalized to be JSON-safe and stripped of empty entries. |
| **`create_orientation_identifier(eulers, hash_length=5, decimals=6)`** | `functions/mimedo_identifier.py` | A phase's Euler-angle array `(n_grains, 3)` as `[Phi1, Phi, Phi2]`, rounded to `decimals`. |
| **`create_microstructure_identifier(...)`** | **Kanapy's own `api.py`** (moved there upstream) | The voxelized geometry — grain and voxel volumes, rounded to significant figures. Returns the `S_`-prefixed id. |

```python
from functions.mimedo_identifier import (
    create_mimedo_identifier, create_orientation_identifier,
)
```

> The `mandatory_fields` whitelist is currently applied to `microstructure` but deliberately
> **not yet** to `"phase"` — a flagged preventive follow-up, not an active bug.
